# Vector Search

### Imports

```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_chroma import Chroma

import pandas as pd
```

### Vector Search

```python
#Convert csv to txt
books = pd.read_csv("../data/books_cleaned.csv")
books["tagged_description"].to_csv("../data/tagged_description.txt", sep="\n", index=False, header=False)

books["tagged_description"].head()
```

```python
#Document Loading and Splitting

raw_documents = TextLoader("../data/tagged_description.txt", encoding="utf-8").load()

text_splitter = CharacterTextSplitter(chunk_size=1, chunk_overlap=0, separator="\n")

documents = text_splitter.split_documents(raw_documents)
```

```python
#Vector database initialization

# Initialize a local open-source embedding model
embedding_model = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Create the Chroma DB using your local embeddings
db_books = Chroma.from_documents(documents, embedding_model)
```
